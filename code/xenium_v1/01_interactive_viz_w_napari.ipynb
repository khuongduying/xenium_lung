## Load cell boundaries

```python
import pandas as pd

cells = pd.read_parquet(
    "../data/Xenium_V1_humanLung_Cancer_FFPE_outs/cell_boundaries.parquet"
)

print(cells.head())
print(cells.columns)
```

```python
import numpy as np

polygons = []

for cell_id, df in cells.groupby("cell_id"):

    polygon = df[
        ["vertex_y", "vertex_x"]
    ].to_numpy()

    polygons.append(polygon)
```

```python
viewer.add_shapes(
    polygons,
    shape_type="polygon",
    edge_width=1,
    face_color=[0, 0, 0, 0],
    edge_color="white",
    name="cell_boundaries"
)
```

```python
import napari

# 1. Initialize the viewer properly
viewer = napari.Viewer()

# 2. Add your layers
viewer.add_shapes(
    polygons,
    shape_type="polygon",
    edge_width=1,
    face_color=[0, 0, 0, 0],
    edge_color="white",
    name="cell_boundaries"
)

napari.run()
```

```python
import os
# Force Qt and OpenGL to use X11/GLX instead of Wayland
os.environ["QT_QPA_PLATFORM"] = "xcb"
os.environ["PYOPENGL_PLATFORM"] = "glx"

import napari

# 1. Initialize the viewer
viewer = napari.Viewer()

# 2. Use the built-in plugin to open the Zarr store directly
viewer.open("data/xenium_sample.zarr", plugin="napari-spatialdata")

# 3. Start the interactive GUI
napari.run()
```

```python
import spatialdata as sd

# 1. Load your existing sdata object if not already in memory
sdata = sd.SpatialData.read("data/xenium_sample.zarr")

# 2. Fix the cell_boundaries index
boundaries = sdata.shapes["cell_boundaries"]
boundaries["cell_id_str"] = boundaries.index.astype(str) # preserve original IDs
boundaries = boundaries.reset_index(drop=True)             # convert index to integers 0, 1, 2...
sdata.shapes["cell_boundaries"] = boundaries

# 3. (Optional) Do the exact same for nucleus_boundaries if needed
if "nucleus_boundaries" in sdata.shapes:
    nuclei = sdata.shapes["nucleus_boundaries"]
    nuclei["cell_id_str"] = nuclei.index.astype(str)
    nuclei = nuclei.reset_index(drop=True)
    sdata.shapes["nucleus_boundaries"] = nuclei

# 4. Overwrite/save your updated Zarr store
sdata.write("data/xenium_sample.zarr", overwrite=True)
```
