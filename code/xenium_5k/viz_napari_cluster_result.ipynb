```python
import pandas as pd
import spatialdata as sd
from napari_spatialdata import Interactive

sdata = sd.SpatialData.read("../../data/xenium_prime_5k_processed.zarr")
adata = sdata["table"]

# Re-attach cluster labels loaded from the CSV (e.g. if adata doesn't already have "leiden")
clustering_res = pd.read_csv("../../data/table/xenium_prime_5k_clustering.csv")
clustering_res = clustering_res.set_index("cell_id")
adata.obs["group"] = clustering_res.loc[adata.obs["cell_id"], "group"].astype("category").values
```

```python
interactive = Interactive(sdata)
interactive.run()
```
